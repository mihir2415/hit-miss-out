```python
import os
import json
import pandas as pd

folder_path = 'all_json'

team_players = {}

for filename in os.listdir(folder_path):
    if filename.endswith('.json'):
        with open(os.path.join(folder_path, filename)) as f:
            data = json.load(f)

        teams = data['info']['teams']  # e.g. ['Mumbai Indians', 'CSK']

        for team in teams:
            players = data['info']['players'].get(team, [])

            if team not in team_players:
                team_players[team] = set()

            team_players[team].update(players)

# Convert to a flat DataFrame
records = [
    {'team': team, 'player': player}
    for team, players in team_players.items()
    for player in players
]

df = pd.DataFrame(records).reset_index(drop=True)
```

```python
df.to_csv('team_players.csv', index=False)
```
